```sql
-- Sprint 4 checkpoint: schema -> volume -> file -> Delta table -> query.
-- Run in Databricks SQL Editor (Free Edition, catalog `workspace`).

CREATE SCHEMA IF NOT EXISTS workspace.crypto;
CREATE VOLUME IF NOT EXISTS workspace.crypto.raw;

-- Upload btc_sample.csv to /Volumes/workspace/crypto/raw/samples/ first (Catalog UI).

CREATE OR REPLACE TABLE workspace.crypto.btc_prices_raw AS
SELECT
  symbol,
  CAST(price_date AS DATE)     AS price_date,
  CAST(close_price AS DOUBLE)  AS close_price
FROM read_files(
  '/Volumes/workspace/crypto/raw/samples/btc_sample.csv',
  format => 'csv',
  header => true
);

SELECT * FROM workspace.crypto.btc_prices_raw ORDER BY price_date;   -- expect 3 rows

DESCRIBE DETAIL workspace.crypto.btc_prices_raw;   -- format = delta, see its storage location
DESCRIBE HISTORY workspace.crypto.btc_prices_raw;  -- the Delta transaction log, version 0
```
